### Reglas de los nombres de las Variables
- Los nombres no deben tener espacios en blanco
- Los nombres no inician con caracteres numericos
- Los nombres no poseen el caracter / o \
- Los nombres no deben contener caraxteres compuestos (caracteres tildados , dieresis , eñe)
- Los nombres debes ser autodocumentados , tener concordancia con el uso que va a tener

### Datos de Naturaleza numerica 
- int
- float

Para variables de naturaleza numerica (int , float) entonces las 4 operaciones numericas son naturales ( + , - , * , / , ** )

Algoritmo de Euclides : D = d * q + r

En pyhton
- q = D // d
- r = D % d

```python
print(var_entrera + var_punto_flotante)
print(var_entrera - var_punto_flotante)
print(var_entrera * var_punto_flotante)
print(var_entrera / var_punto_flotante)
print(var_entrera // var_punto_flotante)
print(var_entrera % var_punto_flotante)
```
